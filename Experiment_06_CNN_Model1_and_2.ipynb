```python
# CNN
import numpy as np
import tensorflow as tf
from tensorflow.keras.datasets import mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, Flatten, Conv2D, MaxPooling2D, AvgPool2D
import matplotlib.pyplot as plt
```

```python
(train_imgs, train_labels), (test_imgs, test_labels) = mnist.load_data()

train_imgs, test_imgs = train_imgs / 255.0, test_imgs / 255.0
```

```python
class_names = ['zero', 'two', 'two', 'three', 'four', 'five', 'six', 'seven', 'eight', 'nine']
```

```python
plt.figure(figsize=(10,10))
for i in range(25):
    plt.subplot(5,5,i+1)
    plt.xticks([])
    plt.yticks([])
    plt.grid(False)
    plt.imshow(train_imgs[i])
    plt.xlabel(class_names[train_labels[i]])
plt.show()
```

```python
model = Sequential([
    Conv2D(32, (3, 3), activation='relu', input_shape=(28, 28, 1)),
    MaxPooling2D((2, 2)),
    Conv2D(64, (3, 3), activation='relu'),
    MaxPooling2D((2, 2)),
    Conv2D(64, (3, 3), activation='relu'),
    Flatten(),
    Dense(64, activation='relu'),
    Dense(10, activation='softmax')]
)
model.summary()
```

```python
model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

history = model.fit(train_imgs, train_labels, epochs=10,
                    validation_data=(test_imgs, test_labels))
```

```python
plt.plot(history.history['accuracy'], label='accuracy')
plt.plot(history.history['val_accuracy'], label = 'val_accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.ylim([0.5, 1])
plt.legend(loc='lower right')

test_loss, test_acc = model.evaluate(test_imgs,  test_labels, verbose=2)
```

```python
model2 = Sequential([
    Conv2D(150, (3, 3), activation='relu', input_shape=(28, 28, 1)),
    AvgPool2D((2, 2)),
    Conv2D(200, (3, 3), activation='relu'),
    AvgPool2D((2, 2)),
    Conv2D(120, (3, 3), activation='relu'),
    AvgPool2D((2, 2)),
    Flatten(),
    Dense(100, activation='relu'),
    Dense(200, activation='relu'),
    Dense(150, activation='relu'),
    Dense(10, activation='softmax')]
)
model2.summary()
```

```python
model2.compile(optimizer='adam',
               loss='sparse_categorical_crossentropy',
               metrics=['accuracy'])

history2 = model2.fit(train_imgs, train_labels, epochs=10, batch_size=32 , validation_data=(test_imgs, test_labels))
```

```python
plt.plot(history2.history['accuracy'], label='accuracy')
plt.plot(history2.history['val_accuracy'], label = 'val_accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.ylim([0.5, 1])
plt.legend(loc='lower right')

test_loss, test_acc = model2.evaluate(test_imgs,  test_labels, verbose=2)
```

```python
import numpy as np

def plot_predictions(model_obj, title, num_images=5):
    predictions = model_obj.predict(test_imgs[:num_images])
    plt.figure(figsize=(15, 3))
    plt.suptitle(title, fontsize=16)
    for i in range(num_images):
        plt.subplot(1, num_images, i + 1)
        plt.xticks([])
        plt.yticks([])
        plt.grid(False)
        plt.imshow(test_imgs[i], cmap=plt.cm.binary)

        predicted_label = np.argmax(predictions[i])
        true_label = test_labels[i]

        color = 'blue' if predicted_label == true_label else 'red'
        plt.xlabel(f"Pred: {class_names[predicted_label]}\nTrue: {class_names[true_label]}", color=color)
    plt.show()

plot_predictions(model, "Model 1 Predictions")

plot_predictions(model2, "Model 2 Predictions")
```

```python
model3 = Sequential([
    Conv2D(6, (5, 5), activation='relu', padding='same', input_shape=(28, 28, 1)),
    MaxPooling2D((2, 2)),
    Conv2D(16, (5, 5), activation='relu', padding='valid'),
    MaxPooling2D((2, 2)),
    Flatten(),
    Dense(120, activation='relu'),
    Dense(84, activation='relu'),
    Dense(10, activation='softmax')]
)
model3.summary()
```

```python
model3.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

history = model3.fit(train_imgs, train_labels, epochs=10,
                    validation_data=(test_imgs, test_labels))
```

```python
plt.plot(history.history['accuracy'], label='accuracy')
plt.plot(history.history['val_accuracy'], label = 'val_accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.ylim([0.5, 1])
plt.legend(loc='lower right')

test_loss, test_acc = model3.evaluate(test_imgs,  test_labels, verbose=2)
```
