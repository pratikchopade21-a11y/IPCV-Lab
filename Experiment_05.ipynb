```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image_path = "input.jpg"  # Replace with your image path
image = cv2.imread("/content/Parrot.jpeg")

if image is None:
    print("Error: Image not found!")
    exit()
```

```python
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
```

```python

sobel_x = cv2.Sobel(gray, cv2.CV_64F, 1, 0, ksize=3)
```

```python
sobel_y = cv2.Sobel(gray, cv2.CV_64F, 0, 1, ksize=3)
```

```python
sobel_magnitude = np.sqrt(sobel_x**2 + sobel_y**2)
```

```python
sobel_magnitude = np.sqrt(sobel_x**2 + sobel_y**2)
```

```python
# Apply Gaussian Blur to reduce noise
blurred = cv2.GaussianBlur(gray, (5, 5), 0)
```

```python
# Canny Edge Detector
canny_edges = cv2.Canny(blurred, threshold1=50, threshold2=150)
```

```python
plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.title("Original Image")
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.subplot(2, 2, 2)
plt.title("Sobel Magnitude")
plt.imshow(sobel_magnitude, cmap='gray')
plt.axis('off')

plt.subplot(2, 2, 3)
plt.title("Canny Edges")
plt.imshow(canny_edges, cmap='gray')
plt.axis('off')

plt.subplot(2, 2, 4)
plt.title("Grayscale Image")
plt.imshow(gray, cmap='gray')
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
# Function: LoG Filter (Laplacian of Gaussian)
# ---------------------------------------
def log_filter(gray):
    # Step 1: Apply Gaussian Blur
    blurred = cv2.GaussianBlur(gray, (5, 5), 0)

    # Step 2: Apply Laplacian
    log = cv2.Laplacian(blurred, cv2.CV_64F)

    # Convert to absolute values
    log = cv2.convertScaleAbs(log)

    return log
```

```python
from google.colab.patches import cv2_imshow

# Display Function
# ---------------------------------------
def show_images(images):
    for title, img in images.items():
        cv2_imshow(img)

    print("Press any key to exit...")
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```
